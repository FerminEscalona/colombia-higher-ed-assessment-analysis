# 1.0 Entendimiento del negocio — Saber Pro 2024

## Propósito del notebook

Este documento desarrolla la fase **Business Understanding** de CRISP-DM para el conjunto `Examen_Saber_Pro_Genericas_2024.txt`. Busca comprender el examen, los actores, el valor del dato, la unidad de análisis, el significado de sus 90 variables y las precauciones necesarias antes de cualquier análisis.

> Este notebook **no es un EDA**. No calcula distribuciones, faltantes, correlaciones ni atípicos; tampoco prepara datos o construye modelos.

## 1. ¿Qué es Saber Pro?

El **Examen de Estado de la Calidad de la Educación Superior Saber Pro** es una evaluación estandarizada administrada por el ICFES. Su población principal son estudiantes próximos a culminar programas profesionales universitarios. La institución de educación superior (IES) debe preregistrar a quienes hayan aprobado por lo menos el 75% de los créditos del programa; los graduados también pueden inscribirse de manera independiente.

Según el ICFES, sus objetivos son:

1. Comprobar el desarrollo de las competencias de estudiantes próximos a culminar el pregrado.
2. Producir información que permita comparar programas, instituciones y metodologías, observar su evolución y construir indicadores de valor agregado.
3. Aportar información para evaluar la calidad de programas, instituciones y del servicio educativo, apoyar procesos de mejoramiento y orientar decisiones de política pública.

El examen combina **competencias genéricas**, comunes a estudiantes de diferentes áreas, y **competencias específicas**, asociadas con áreas de formación. Este archivo contiene resultados de las cinco competencias genéricas: Lectura Crítica, Razonamiento Cuantitativo, Competencias Ciudadanas, Comunicación Escrita e Inglés.

## 2. Problema y oportunidad de negocio

En este proyecto, «negocio» no significa una actividad comercial: representa las necesidades de información del sistema colombiano de educación superior.

### Necesidad central

Comprender cómo se comportan los resultados de competencias genéricas de Saber Pro 2024 y cómo se relacionan descriptivamente con características del estudiante, su hogar, el programa y la institución.

### Valor esperado

El dataset puede apoyar:

- la identificación de fortalezas y oportunidades de mejoramiento curricular;
- la descripción de brechas asociadas con contextos socioeconómicos, territoriales o institucionales;
- la comparación contextualizada entre módulos, programas, núcleos de formación e instituciones;
- el seguimiento de indicadores de calidad;
- la formulación de preguntas para análisis posteriores;
- decisiones institucionales y de política educativa basadas en evidencia.

### Lo que el dataset no responde por sí solo

No permite afirmar que una característica **cause** un resultado. Tampoco mide por sí solo toda la calidad de una institución o programa: el aprendizaje universitario es multidimensional y los puntajes dependen, entre otros factores, del contexto previo, la composición de la población, la motivación y las condiciones de aplicación.

## 3. Actores y decisiones que pueden apoyarse

| Actor | Interés principal | Decisiones o usos posibles |
|---|---|---|
| ICFES y Ministerio de Educación | Calidad y funcionamiento del sistema | Indicadores, seguimiento, política pública e investigación |
| Instituciones de educación superior | Desempeño de sus estudiantes y programas | Mejoramiento curricular, acompañamiento y autoevaluación |
| Directivos de programa | Fortalezas y debilidades por competencia | Priorización de acciones académicas |
| Docentes | Competencias que requieren refuerzo | Diseño de estrategias de enseñanza |
| Estudiantes y graduados | Lectura de su resultado | Reconocimiento de fortalezas y áreas de desarrollo |
| Investigadores | Equidad, calidad y resultados educativos | Estudios descriptivos, comparativos o explicativos |
| Sociedad | Rendición de cuentas sobre educación superior | Consulta de resultados agregados y tendencias |

Las comparaciones deben considerar el programa, el Núcleo Básico del Conocimiento (NBC), la metodología, el territorio y la composición de los estudiantes. Un ranking simple de puntajes puede producir interpretaciones injustas.

## 4. Alcance del activo de datos

- **Archivo:** `data/raw/Examen_Saber_Pro_Genericas_2024.txt`.
- **Tema:** resultados individuales y variables de contexto de Saber Pro 2024 para competencias genéricas.
- **Unidad de análisis esperada:** una presentación o registro de una persona evaluada.
- **Identificador técnico:** `estu_consecutivo`; debe tratarse como una llave, no como una característica explicativa.
- **Cobertura conceptual:** estudiante, contexto familiar, programa académico, institución, lugar de presentación y resultados.
- **Resultados disponibles:** puntaje global y resultados por los cinco módulos genéricos.
- **Fuera de alcance:** respuestas por pregunta, notas universitarias, calidad docente directa, trayectorias laborales y resultados de módulos específicos.

### Tipos de registro

`estu_estudiante` diferencia registros institucionales (`ESTUDIANTE`) de inscripciones independientes (`INDIVIDUAL`). El ICFES indica que los graduados pueden inscribirse independientemente y que su resultado no afecta los resultados agregados. `estu_agregado` funciona como indicador de inclusión o elegibilidad para agregación; debe respetarse cuando se construyan resultados institucionales.

## 5. Cómo leer los nombres de las variables

Los prefijos permiten reconocer el tema de cada columna:

| Prefijo | Tema |
|---|---|
| `estu_` | Estudiante, inscripción, ubicación o programa |
| `fami_` | Contexto familiar y del hogar |
| `inst_` | Institución de educación superior |
| `mod_` | Resultado de un módulo del examen |

Los sufijos de resultados significan:

| Sufijo | Interpretación |
|---|---|
| `_punt` | Puntaje obtenido en el módulo |
| `_desem` | Nivel de desempeño alcanzado |
| `_pnal` | Percentil frente al grupo nacional de comparación |
| `_pnbc` | Percentil frente al Núcleo Básico del Conocimiento |

Un **puntaje** expresa desempeño en la escala del examen. Un **percentil** expresa posición relativa: por ejemplo, un percentil 80 indica una posición igual o superior a la de aproximadamente el 80% del grupo de referencia. No significa 80% de respuestas correctas. Un **nivel de desempeño** agrupa puntajes en categorías interpretables de dominio.

## 13. Criterios de éxito del proyecto

### Éxito analítico

- Interpretar correctamente puntajes, percentiles, niveles y grupos de comparación.
- Separar identificadores y códigos de las variables medibles.
- Reportar resultados con el contexto de programa, NBC, metodología e institución.
- Hacer visibles los faltantes, límites y sesgos de cobertura.
- Evitar afirmaciones causales basadas únicamente en asociaciones.

### Éxito para el negocio educativo

- Producir hallazgos comprensibles y accionables para actores educativos.
- Evitar rankings descontextualizados o estigmatización de estudiantes e instituciones.
- Permitir que las decisiones puedan rastrearse hasta variables y definiciones documentadas.
- Diferenciar claramente desempeño observado, contexto y posibles hipótesis.

## 14. Riesgos, ética y uso responsable

### Privacidad

Aunque `estu_consecutivo` no sea un nombre, su combinación con fecha de nacimiento, institución, programa, municipio, discapacidad o etnia puede facilitar reidentificación. Los resultados deben divulgarse de forma agregada y con tamaños mínimos de grupo.

### Equidad

Discapacidad, etnia, género, territorio y nivel socioeconómico son variables sensibles. Sirven para estudiar brechas y acceso a oportunidades, no para justificar trato desfavorable ni decisiones automáticas sobre personas.

### Comparabilidad

- Un percentil es relativo al grupo de referencia.
- El percentil nacional y el percentil NBC responden preguntas diferentes.
- Las comparaciones entre instituciones deben controlar diferencias de composición y cobertura.
- Los registros `INDIVIDUAL` y aquellos no incluidos en agregados requieren tratamiento conceptual separado.
- Si se comparan años, primero debe verificarse la estabilidad de escalas, población, módulos y metodología.

### Calidad de captura

Varias variables provienen de autorreporte y pueden contener omisiones, categorías ambiguas o errores. En particular, `estu_repite`, los campos de cursos de preparación y las parejas `ocupacion`/`trabajolabor` requieren validación documental antes de un uso sustantivo.

## 15. Supuestos y límites de esta documentación

- Se interpreta cada fila como un registro de presentación; la unicidad efectiva debe verificarse en Data Understanding.
- Las definiciones detalladas de campos administrativos pueden cambiar entre entregas. Ante una discrepancia, prevalece el diccionario de datos oficial correspondiente al archivo 2024.
- Las descripciones de variables con nombres abreviados son operativas y no sustituyen el formulario de inscripción o diccionario oficial.
- Este archivo contiene competencias genéricas; no debe generalizarse a módulos específicos que no están presentes.
- La ausencia de una respuesta no significa necesariamente que la característica no exista; puede deberse al diseño del formulario, al tipo de inscripción o a falta de reporte.
